# Apache Airflow 3.3.2 – Notes & Command Cheat Sheet

## 1. Start Airflow

```bash
# Navigate to project
cd ~/Documents/VS/Air_flow

# Activate virtual environment
source airflow_env/bin/activate

# Start Airflow
airflow standalone
```

**Airflow UI:** http://localhost:8080

**Stop Airflow:** `Ctrl + C`

## 2. Basic Commands

| Command | Description |
|---|---|
| `airflow version` | Check Airflow version |
| `airflow info` | Airflow environment information |
| `airflow --help` | Display all commands |
| `airflow cheat-sheet` | Display built-in cheat sheet |

## 3. DAG Commands

**DAG = Directed Acyclic Graph (workflow)**

```bash
airflow dags list                         # List all DAGs
airflow dags list-import-errors           # Check import errors
airflow dags details sales_pipeline       # Show DAG details
airflow dags show sales_pipeline          # Show DAG structure
airflow dags pause sales_pipeline         # Pause DAG
airflow dags unpause sales_pipeline       # Resume DAG
airflow dags trigger sales_pipeline       # Trigger DAG manually
airflow dags list-runs sales_pipeline     # List DAG runs
airflow dags state sales_pipeline RUN_ID  # Check DAG run status
```

## 4. Task Commands

**Task = Individual unit of work inside a DAG**

```bash
airflow tasks list sales_pipeline
# List tasks in a DAG

airflow tasks test sales_pipeline extract_data 2026-11-02
# Test an individual task

airflow tasks state sales_pipeline extract_data RUN_ID
# Check task status

airflow tasks states-for-dag-run sales_pipeline RUN_ID
# Check all task statuses in a DAG run

airflow tasks failed-deps sales_pipeline extract_data RUN_ID
# Check unmet dependencies

airflow tasks render sales_pipeline extract_data RUN_ID
# View rendered templates
```

## 5. Testing & Debugging

```bash
airflow dags test sales_pipeline 2026-11-02
# Test the complete DAG

airflow dags list-import-errors
# Find DAG import errors

airflow tasks clear sales_pipeline \
  --start-date 2026-11-02 \
  --end-date 2026-11-02
# Clear task states so tasks can run again
```

## 6. Database Commands

```bash
airflow db check       # Check database connection
airflow db migrate     # Initialize or migrate database
airflow db reset       # Reset database (DANGER: deletes metadata)
```

## 7. Variables Commands

**Variables = Store reusable values**

```bash
airflow variables list
airflow variables set input_file sales.csv
airflow variables get input_file
airflow variables delete input_file
airflow variables export variables.json
airflow variables import variables.json
```

## 8. Connections Commands

**Connections = Connect Airflow to external systems**

```bash
airflow connections list
airflow connections get my_connection
airflow connections test my_connection
airflow connections delete my_connection
```

## 9. Configuration Commands

```bash
airflow config list
# List configuration options

airflow config get-value core dags_folder
# Find DAG folder

airflow config get-value core executor
# Check executor

airflow info
# View environment information
```

Open DAG folder in Finder:

```bash
open "$(airflow config get-value core dags_folder)"
```

## 10. Provider Commands

**Providers = Integrations with external tools and services**

```bash
airflow providers list
# List installed providers

python -m pip list
# List installed Python packages

python -m pip show apache-airflow-providers-postgres
# Check PostgreSQL provider
```

## 11. Important Airflow Terms

| Term | Meaning |
|---|---|
| DAG | Defines the workflow |
| Task | Individual operation in a workflow |
| DAG Run | One execution of a DAG |
| Operator | Defines what a task does |
| Scheduler | Determines when tasks should run |
| Executor | Determines how tasks are executed |
| Metadata Database | Stores workflow and execution information |
| Web UI | Interface to monitor workflows |

## 12. Daily Workflow

```bash
# 1. Navigate to project
cd ~/Documents/VS/Air_flow

# 2. Activate environment
source airflow_env/bin/activate

# 3. Start Airflow
airflow standalone

# 4. Check DAGs
airflow dags list

# 5. Check errors
airflow dags list-import-errors

# 6. Test DAG
airflow dags test DAG_ID 2026-11-02

# 7. Unpause DAG
airflow dags unpause DAG_ID

# 8. Trigger DAG
airflow dags trigger DAG_ID

# 9. Check DAG runs
airflow dags list-runs DAG_ID
```

## Important Notes

- `DAG_ID` = Your actual DAG ID.
- `TASK_ID` = Your actual task ID.
- `RUN_ID` = Actual DAG run ID.
- Replace example dates with your required logical date.
- `airflow dags test` tests a DAG run.
- `airflow tasks test` tests an individual task without recording its state in the metadata database.
- Clearing task states allows tasks to execute again.
- Do not use `airflow db reset` unless you intend to delete existing metadata.
- You do not need to reinstall Airflow for every new project.
